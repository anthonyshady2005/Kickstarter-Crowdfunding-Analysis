The column 'usd pledged' has many incorrect values so we decided to drop it completely

```python
df_clean=df.copy()
df_clean.drop('usd pledged',axis=1,inplace=True)
df_clean.info()
```

The dataset has only 4 missing values for the column 'name' which is negligible so we decided to drop the rows with those missing values.

```python
df_clean.dropna(inplace=True)
df_clean.info()
```

```python
print(df_clean.duplicated().sum())

# No duplicates were found
```

```python
print(type(df_clean.loc[0,"deadline"]))
print(type(df_clean.loc[0,"launched"]))
```

The 'deadline' and 'launched' columns have string data type instead of date, so we changed their types.

```python
df_clean['deadline']=pd.to_datetime(df_clean['deadline'])
df_clean['launched']=pd.to_datetime(df_clean['launched'])
df_clean.head()
df_clean.info()
```

```python
df_clean['state'].unique()
```

```python
df_clean=df_clean[df_clean['state']!='live']
```

```python
df_clean['CampaignDuration']=(df_clean['deadline']-df_clean['launched']).dt.days
df_clean['FundingRatio']=df_clean['usd_pledged_real']/df_clean['usd_goal_real']

def categorize_goal(goal):
    if goal < 10000:
        return "Low"
    elif 10000 <= goal <= 100000:
        return "Medium"
    else:
        return "High"
df_clean["GoalCategory"] = df_clean["usd_goal_real"].apply(categorize_goal)

df_clean['LaunchMonth']=df_clean['launched'].dt.month
df_clean.head()
```

```python
session.execute("""CREATE TABLE kickstarter_projects (
    id bigint PRIMARY KEY,
    name text,
    category text,
    main_category text,
    currency text,
    deadline timestamp,
    goal double,
    launched timestamp,
    pledged double,
    state text,
    backers int,
    country text,
    usd_pledged_real double,
    usd_goal_real double,
    campaign_duration int,
    funding_ratio double,
    goal_category text,
    launch_month int
);""")
```

```python
insert_query = session.prepare("""
INSERT INTO kickstarter_projects (
id, name, category, main_category, currency,
deadline, goal, launched, pledged, state,
backers, country, usd_pledged_real, usd_goal_real,
campaign_duration, funding_ratio, goal_category, launch_month
)
VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
""")
```

```python
from cassandra.query import BatchStatement

batch_size = 300

for start in range(0, len(df_clean), batch_size):

    batch = BatchStatement()
    chunk = df_clean.iloc[start:start+batch_size]

    for _, row in chunk.iterrows():

        batch.add(insert_query, (
            int(row.ID),
            str(row.name),
            str(row.category),
            str(row.main_category),
            str(row.currency),
            row.deadline,
            float(row.goal),
            row.launched,
            float(row.pledged),
            str(row.state),
            int(row.backers),
            str(row.country),
            float(row.usd_pledged_real),
            float(row.usd_goal_real),
            int(row.CampaignDuration),
            float(row.FundingRatio),
            str(row.GoalCategory),
            int(row.LaunchMonth)
        ))

    session.execute(batch)

    print("Inserted rows:", start + len(chunk))
```

```python
# What is the success rate for each main category?

rows = session.execute("""
SELECT main_category, state
FROM kickstarter_projects
""")

df_cassandra = pd.DataFrame(list(rows))

success_rate = (
    df_cassandra.groupby("main_category")["state"]
    .apply(lambda x: (x == "successful").mean())
    .reset_index(name="success_rate")
)

print(success_rate)
```

```python
#  Which countries have the highest average pledged amount in USD?

rows = session.execute("""
SELECT country, usd_pledged_real
FROM kickstarter_projects
""")

df_cassandra = pd.DataFrame(list(rows))

avg_pledged = (
    df_cassandra.groupby("country")["usd_pledged_real"]
    .mean()
    .sort_values(ascending=False)
)

print(avg_pledged)
```

```python
# Among failed projects, which category has the highest number of failures?

rows = session.execute("""
SELECT id, name, funding_ratio
FROM kickstarter_projects
""")

df_cassandra = pd.DataFrame(list(rows))

top_projects = (
    df_cassandra.sort_values("funding_ratio", ascending=False)
    [["id", "name", "funding_ratio"]]
    .head(5)
)

print(top_projects)
```

```python
# Among failed projects, which category has the highest number of failures?

rows = session.execute("""
SELECT category, state
FROM kickstarter_projects
""")

df_cassandra = pd.DataFrame(list(rows))

failed_categories = (
    df_cassandra[df_cassandra["state"] == "failed"]
    .groupby("category")
    .size()
    .sort_values(ascending=False)
)

print(failed_categories.head(10))
```

```python
# What is the average campaign duration for successful vs failed projects?

rows = session.execute("""
SELECT state, campaign_duration
FROM kickstarter_projects
""")

df_cassandra = pd.DataFrame(list(rows))

avg_duration = (
    df_cassandra.groupby("state")["campaign_duration"]
    .mean()
)

print(avg_duration)
```

```python
# Which month has the highest number of successful projects?

rows = session.execute("""
SELECT launch_month, state
FROM kickstarter_projects
""")

df_cassandra = pd.DataFrame(list(rows))

successful_month = (
    df_cassandra[df_cassandra["state"] == "successful"]
    .groupby("launch_month")
    .size()
    .sort_values(ascending=False)
)

print(successful_month)
```

```python
# What is the average number of backers per main category?

rows = session.execute("""
SELECT main_category, backers
FROM kickstarter_projects
""")

df_cassandra = pd.DataFrame(list(rows))

avg_backers = (
    df_cassandra.groupby("main_category")["backers"]
    .mean()
    .sort_values(ascending=False)
)

print(avg_backers)
```

```python
# What percentage of projects reached or exceeded their goal?

rows = session.execute("""
SELECT funding_ratio
FROM kickstarter_projects
""")

df_cassandra = pd.DataFrame(list(rows))

goal_percentage = (df_cassandra["funding_ratio"] >= 1).mean() * 100

print("Percentage reaching goal:", goal_percentage)
```
